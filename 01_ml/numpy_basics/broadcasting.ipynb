#numpy基础知识

```python
import numpy as np
import copy
```

```python
A=np.arange(12).reshape(3,4)
```

```python
print(A)
```

##赋值

```python
B=A
B[0,0]=99
print(A)
np.shares_memory(A,B)
```

##拷贝

```python
A=np.arange(12).reshape(3,4)
B=A.copy()
B[1,1]=99
print(A)
np.shares_memory(A,B)
```

##切片

```python
A=np.arange(12).reshape(3,4)
B=A[0:2]
B[0,0]=99
print(A)
np.shares_memory(A,B)
```

##数字*和矩阵@ 

```python
A=np.arange(4).reshape(-1,2)
B=np.arange(4).reshape(2,2)
print(A*B)
print(A@B)
```

##广播

```python
C=np.arange(2).reshape(2,)
display(C)
print(A+C)
```

##判断这两个广播后的格式

```python
np.broadcast_shapes((3,4),(4,))
```
